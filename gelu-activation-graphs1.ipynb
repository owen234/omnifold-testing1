```python
import matplotlib.pyplot as plt
import numpy as np
import scipy.special
```

```python
npoints = 100

xvals = np.linspace(-3,3,npoints)

yvals = xvals * 0.5 * (1 + scipy.special.erf(xvals/np.sqrt(2)))
yvals2 = xvals * 0.5 * (1 + scipy.special.erf(2*xvals/np.sqrt(2)))
yvals5 = xvals * 0.5 * (1 + scipy.special.erf(5*xvals/np.sqrt(2)))


```

```python
fig,ax = plt.subplots( 1, 1, figsize=(18,12))

axi = ax

axi.plot(xvals,yvals)
axi.plot(xvals,yvals2)
axi.plot(xvals,yvals5)
axi.grid()

plt.show
```

```python
yvals_tanh = 0.5 * xvals * (1 + np.tanh( np.sqrt(2./3.14159)*(xvals+0.044715*np.power(xvals,3))))
```

```python
fig,ax = plt.subplots( 1, 1, figsize=(18,12))

axi = ax

axi.plot(xvals,yvals, linewidth=5)
axi.plot(xvals,yvals_tanh, linestyle='dashed')

axi.grid()

plt.show
```

```python
tanh1 = np.tanh( np.sqrt(2./3.14159)*(xvals+0.044715*np.power(xvals,3)))
```

```python
plt.plot(xvals, tanh1)
```

```python
tanh2 = np.tanh( 2* np.sqrt(2./3.14159)*(xvals+0.044715*np.power(xvals,3)))
tanh5 = np.tanh( 5* np.sqrt(2./3.14159)*(xvals+0.044715*np.power(xvals,3)))
```

```python
plt.plot(xvals, tanh1)
plt.plot(xvals, tanh2)
plt.plot(xvals, tanh5)
```

```python
yvals_tanh1 = 0.5 * xvals * (1 + np.tanh( np.sqrt(2./3.14159)*(xvals+0.044715*np.power(xvals,3))))
yvals_tanh2 = 0.5 * xvals * (1 + np.tanh( 2 * np.sqrt(2./3.14159)*(xvals+0.044715*np.power(xvals,3))))
yvals_tanh5 = 0.5 * xvals * (1 + np.tanh( 5 * np.sqrt(2./3.14159)*(xvals+0.044715*np.power(xvals,3))))
```

```python
fig,ax = plt.subplots( 1, 1, figsize=(9,6))

lw = 2

axi = ax

axi.plot(xvals,yvals_tanh1, linewidth=lw, label='beta = 1')
axi.plot(xvals,yvals_tanh2, linewidth=lw, label='beta = 2')
axi.plot(xvals,yvals_tanh5, linewidth=lw, label='beta = 5')

axi.legend( fontsize=16 )

axi.grid()

plt.show
```
